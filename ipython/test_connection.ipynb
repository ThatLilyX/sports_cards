```python
import uuid
from IPython.display import display_javascript, display_html, display
import json

class RenderJSON(object):
    def __init__(self, json_data):
        if isinstance(json_data, dict):
            self.json_str = json.dumps(json_data)
        else:
            self.json_str = json_data
        self.uuid = str(uuid.uuid4())

    def _ipython_display_(self):
        display_html('<div id="{}" style="height: 600px; width:100%;"></div>'.format(self.uuid), raw=True)
        display_javascript("""
        require(["https://rawgit.com/caldwell/renderjson/master/renderjson.js"], function() {
        document.getElementById('%s').appendChild(renderjson(%s))
        });
        """ % (self.uuid, self.json_str), raw=True)
```

```python
from ebaysdk.finding import Connection as Finding
from ebaysdk.shopping import Connection as Shopping
```

```python
def finding(keywords, pageNum, execMethod="findItemsAdvanced", **kwargs):
    
    api = Finding(domain="svcs.sandbox.ebay.com", config_file='test_yaml.yaml', siteid="EBAY-US", https=True)
    request = {
            'keywords': keywords,
#             'itemFilter': [
#                 {'name': 'condition', 'value': 'new'}
#             ],
            'paginationInput': {
                'entriesPerPage': 100,
                'pageNumber': pageNum
            },
            'sortOrder': 'PricePlusShippingLowest',
            'descriptionSearch': 'true'
        }
    response = api.execute(execMethod, request)
    return response
```

```python
def shopping(keywords, pageNum, searchMethod="FindProducts"):
    
    api = Shopping(domain="open.api.sandbox.ebay.com", config_file='test_yaml.yaml', siteid="EBAY-US")
    request = {
            'QueryKeywords': keywords,
        }
    response = api.execute(searchMethod, request)
    return response
```

```python
response = shopping("Car", 1)
```
